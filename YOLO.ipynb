```python
from ultralytics import YOLO
import os
import torch

# --------------------------
# 1️⃣ Check if GPU is available
# --------------------------
if torch.cuda.is_available():
    device = 0  # use GPU
    print("✅ GPU detected. Training on GPU.")
else:
    device = 'cpu'
    print("⚠️ No GPU detected. Training on CPU.")

# --------------------------
# 2️⃣ Dataset YAML path
# --------------------------
data_yaml = r"C:/Users/STF8586/OneDrive - University of Derby/Desktop/S.K/yolodataset/data.yaml"

# --------------------------
# 3️⃣ Ensure output directory exists
# --------------------------
output_dir = r"C:/Users/STF8586/OneDrive - University of Derby/Desktop/S.K/output-yolo"
os.makedirs(output_dir, exist_ok=True)

# --------------------------
# 4️⃣ Load YOLOv8 pretrained model
# --------------------------
model = YOLO('yolov8s.pt')  # official checkpoint

# --------------------------
# 5️⃣ Train YOLOv8 on your dataset using GPU
# --------------------------
results = model.train(
    data=data_yaml,    # path to dataset YAML
    epochs=20,         # number of training epochs
    imgsz=640,         # image size
    batch=16,          # batch size
    workers=4,         # data loader workers
    device=device,     # automatically uses GPU if available
    augment=True       # data augmentation
)

# --------------------------
# 6️⃣ Evaluate the trained model
# --------------------------
metrics = model.val(device=device)  # Evaluate using GPU
print(metrics)
```

```python
import random
import os
from ultralytics import YOLO
from PIL import Image

# Load trained model
model = YOLO("C:/Users/STF8586/Downloads/S.K/runs/detect/train8/weights/best.pt")

# Path to your test images folder (from your data.yaml)
test_folder = "C:/Users/STF8586/OneDrive - University of Derby/Desktop/S.K/yolodataset/images/test"

# List all images in the folder
all_images = [os.path.join(test_folder, f) for f in os.listdir(test_folder) 
              if f.lower().endswith(('.jpg', '.jpeg', '.png'))]

# Pick N random images
N = 5  # number of random images to test
sample_images = random.sample(all_images, N)

# Run detection on each random image
for img_path in sample_images:
    results = model.predict(img_path, save=True)  # save=True will save annotated image
    print(f"Processed: {img_path}")
    # Show annotated image in notebook
    annotated_img_path = results[0].plot()  # returns numpy image with bounding boxes
    Image.fromarray(annotated_img_path).show()
```
